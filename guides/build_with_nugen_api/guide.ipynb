### **Build an Application Using Your Aligned Model**

This cookbook guides you through the steps to use your aligned model with the Nugen Inference API and build an application such as a chatbot, RAG application, AI assistant, agent, answer engine, or another application of your choice.

### **Step 1: Complete Alignment and Deploy Your Model**

Complete the model alignment process on the [Nugen Platform](https://platform.nugen.in/).
Once the alignment is successfully completed, deploy your aligned model.
Make sure the deployment is completed successfully before continuing to the next step.

### **Step 2: Get Your Model ID**
1. Open the Model page on the [Nugen Platform](https://platform.nugen.in/).
2. Find the Domain-Aligned Models section.
3. Locate the aligned model that you deployed.
4. Open the model details.
5. Find  # for the Model ID and copy it.
Keep your Model ID ready. You will use it when making inference requests.



### **Step 3: Create Your API Key**
1. Open the API Keys section on the [Nugen platform](https://platform.nugen.in/).
2. Click Create API Key.
3. Create a new API key.
4. Copy the generated API key and keep it safe.


### **Step 4: Open Nugen Documentation**
**Open** [Nugen API Documentation](https://docs.nugen.in )

The Nugen documentation contains the API information and examples required to use your aligned model.



### **Step 5: Go to the Inference Section**
From the documentation, navigate to the Inference section.



### **Step 6: Open Generate Chat Completions**

Select [Generate Chat Completions](https://docs.nugen.in/api-reference/inference/generate-chat-completions).

This page contains the Inference API details and example request for generating chat completions using an aligned model.
Review the API endpoint, request parameters, headers, and example request provided on this page.



### **Step 7: Use the Copy Page Option**

On the [Generate Chat Completions](https://docs.nugen.in/api-reference/inference/generate-chat-completions) page, look for the Copy Page option.
The Copy Page option provides a dropdown with options to open or use the documentation with AI assistants such as ChatGPT or Claude.
You can use this option to provide the Nugen API documentation to an AI coding assistant and get help creating your application.






### **Step 8: Decide What You Want to Build**

You can use your aligned model to build an application of your choice.

For example:
1. Chatbot
2. RAG application
3. Agent
4. AI assistant
4. Document Q&A application
5. Text generation application

Your own application idea

The main requirement is that your application should use the Nugen Inference API to run inference on your aligned model.


### **Step 9: Ask ChatGPT or Claude to Build Your Application**
Use the Copy Page option from the Nugen documentation to provide the API documentation to ChatGPT or Claude.
Then describe what you want to build.



### **Step 10: Use Your Application**

Once your application is ready, start using it with your aligned model by asking queries related to the documents you uploaded.

Ask questions about the document content and interact with your application to get responses from your aligned model.

The goal is to use your ready application with your aligned model and get responses based on your documents.

### **Final Goal**

By the end, you should have:

Your aligned model → inference API → working application

You are free to choose the application you want to build, such as a RAG application, agent, AI assistant, chatbot, or any other application using your aligned model.
